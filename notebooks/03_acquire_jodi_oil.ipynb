# 03 — Acquire JODI Oil secondary products

Download the full JODI secondary-product CSV archive, preserve the raw ZIP, and record its checksum.


```python
from pathlib import Path
import sys

ROOT = Path.cwd()
if not (ROOT / "pyproject.toml").exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / "src"))

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from portugal_refining_resilience.config import get_paths
from portugal_refining_resilience.io import persist_dataframe, write_json

PATHS = get_paths(ROOT)
pd.set_option("display.max_columns", 100)

from portugal_refining_resilience.sources import download_file, load_source_manifest
from portugal_refining_resilience.io import sha256_file
```

```python
sources = load_source_manifest(ROOT / "config" / "sources.yml")
url = sources["jodi_secondary"]["download_url"]
target = PATHS.raw / "jodi" / "world_secondary_csv.zip"
download_file(url, target)
record = pd.DataFrame([{"source": "JODI", "path": str(target.relative_to(ROOT)), "url": url, "sha256": sha256_file(target)}])
persist_dataframe(record, PATHS.provenance / "jodi_download_manifest.csv", key_columns=["source"])
record
```
