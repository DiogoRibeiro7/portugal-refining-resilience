## Research questions

1. How did diesel and gasoline refinery output, imports, exports and domestic demand change from 2005–2024?
2. Did net dependence on imported finished products increase after the Matosinhos transition?
3. Is 2022 statistically exceptional relative to pre-closure variability, especially for diesel?
4. Did Portuguese pre-tax retail fuel prices become more tightly coupled to Spanish comparison-market prices after the transition?
5. Which findings are robust to source choice, event-window choice and the treatment of 2021 as a transition year?
