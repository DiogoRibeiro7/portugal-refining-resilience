# 18 — Final tables, figures and metrics

Create compact report-ready tables from persisted analytical outputs. No numbers should be manually transcribed from charts.


```python
from pathlib import Path
import sys

ROOT = Path.cwd()
if not (ROOT / "pyproject.toml").exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / "src"))

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from portugal_refining_resilience.config import get_paths
from portugal_refining_resilience.io import persist_dataframe, write_json

PATHS = get_paths(ROOT)
pd.set_option("display.max_columns", 100)
```

```python
panel = pd.read_csv(PATHS.processed / "fuel_annual_analytical_panel.csv")
headline_cols = [c for c in ["year", "product", "exports_kt", "imports_kt", "demand_kt", "refinery_output_kt", "net_imports_kt", "net_import_dependence", "domestic_output_coverage", "refining_regime"] if c in panel.columns]
headline = panel.loc[panel["year"].isin([2012, 2013, 2020, 2021, 2022, 2023, 2024]), headline_cols]
headline.to_csv(PATHS.tables / "headline_event_years.csv", index=False)
display(headline)
```

```python
metric_files = sorted(PATHS.metrics.glob("*.csv"))
metric_catalog = pd.DataFrame([{"file": p.name, "rows": len(pd.read_csv(p)), "columns": ", ".join(pd.read_csv(p, nrows=0).columns)} for p in metric_files])
persist_dataframe(metric_catalog, PATHS.metrics / "metric_catalog.csv", key_columns=["file"])
display(metric_catalog)
```
